# Convert JSONL to CSV
Run this cell to convert all `reference_data/*.jsonl` files into `reference_data/*.csv` with headers inferred from JSON keys.

```python
import json, csv, pathlib
folder=pathlib.Path('reference_data')
for p in sorted(folder.glob('*.jsonl')):
    out=folder/(p.stem+'.csv')
    with p.open('r',encoding='utf-8') as f:
        rows=[json.loads(line) for line in f if line.strip()]
    if not rows:
        print('skipped empty', p)
        continue
    # collect columns from union preserving order
    cols=[]
    for r in rows:
        for k in r:
            if k not in cols:
                cols.append(k)
    with out.open('w',encoding='utf-8',newline='') as f:
        w=csv.DictWriter(f, fieldnames=cols)
        w.writeheader()
        w.writerows(rows)
    print('wrote',out)
PY
```
